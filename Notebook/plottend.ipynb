```python
import soundfile as sf
import numpy as np
import matplotlib.pyplot as plt

# 🔧 Parameter
wav_path = "block_0011.wav"  # Pfad zur WAV-Datei
x = 10  # Anzahl der letzten Samples, die du sehen willst

# 📥 WAV laden
data, samplerate = sf.read(wav_path)
print(f"📄 Geladen: {wav_path}")
print(f"🔢 Länge: {len(data)} Samples ({len(data)/samplerate:.2f} Sekunden)")
print(f"🎯 Zeige die letzten {x} Samples")

# Sicherstellen, dass x nicht größer als das Signal ist
if x > len(data):
    raise ValueError(f"x={x} ist größer als die Länge des Signals ({len(data)} Samples)")

# 🧩 Letzten x Samples
last_samples = data[-x:]
time_axis = np.linspace(0, x / samplerate, num=x)

# 📊 Plot
plt.figure(figsize=(10, 3))
plt.plot(time_axis, last_samples)
plt.title(f"Letzte {x} Samples der WAV-Datei")
plt.xlabel("Zeit [s]")
plt.ylabel("Amplitude")
plt.grid(True)
plt.tight_layout()
plt.show()
```
